# Cart Abandonment Prediction — Model Training Pipeline

**Project**: Big Data-Based E-Commerce Cart Abandonment Prediction and Intelligent Customer Retention

**Dataset**: RetailRocket E-commerce (preprocessed)

This notebook:
1. Loads and validates the preprocessed `model_ready_dataset.parquet`
2. Performs a chronological train/validation/test split
3. Trains Logistic Regression, Random Forest, and XGBoost
4. Selects the optimal threshold on the validation set
5. Evaluates the best model on the held-out test set
6. Generates SHAP explanations
7. Exports all deployment artifacts

## 1. Setup & Configuration

```python
# Install required packages (Colab has most pre-installed)
!pip install -q xgboost shap

import pandas as pd
import numpy as np
import json
import joblib
import os
import warnings
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    precision_recall_curve, auc, f1_score, precision_score,
    recall_score, roc_curve, average_precision_score
)
import matplotlib.pyplot as plt
import seaborn as sns
import shap

warnings.filterwarnings('ignore', category=FutureWarning)
sns.set_theme(style='whitegrid')
np.random.seed(42)

# === CONFIGURATION ===
# Upload model_ready_dataset.parquet to Colab, or mount Google Drive
# Option A: Upload directly
# from google.colab import files
# uploaded = files.upload()  # select model_ready_dataset.parquet

# Option B: Google Drive (uncomment and adjust path)
# from google.colab import drive
# drive.mount('/content/drive')
# DATASET_PATH = '/content/drive/MyDrive/your_folder/model_ready_dataset.parquet'

# Option C: Already in /content/
DATASET_PATH = '/content/model_ready_dataset.parquet'

# Output directory for artifacts
ARTIFACTS_DIR = '/content/artifacts'
os.makedirs(ARTIFACTS_DIR, exist_ok=True)

print('Setup complete.')
```

## 2. Load & Validate Dataset

```python
# Load the dataset (supports both Parquet and CSV)
if DATASET_PATH.endswith('.parquet'):
    df = pd.read_parquet(DATASET_PATH)
elif DATASET_PATH.endswith('.csv'):
    df = pd.read_csv(DATASET_PATH)
else:
    raise ValueError(f'Unsupported file format: {DATASET_PATH}')

print(f'Dataset shape: {df.shape}')
print(f'Columns: {list(df.columns)}')
print(f'\nData types:')
print(df.dtypes)
print(f'\nMissing values:')
print(df.isnull().sum())
print(f'\nDuplicate rows: {df.duplicated().sum()}')
```

```python
# === VALIDATION ===
TARGET_COL = 'label'
TIMESTAMP_COL = 'prediction_timestamp'
ID_COLS = ['session_id', 'visitorid', 'itemid']
LEAKAGE_COLS = ['converted']

FEATURE_COLS = [
    'hist_views', 'hist_addtocarts', 'hist_transactions', 'hist_unique_items',
    'session_views', 'session_addtocarts', 'session_unique_items',
    'session_duration_sec', 'hist_conversion_rate',
    'categoryid', 'parent_categoryid', 'num_available_properties'
]

# Verify all expected columns exist
expected_cols = set(FEATURE_COLS + [TARGET_COL, TIMESTAMP_COL] + ID_COLS + LEAKAGE_COLS)
actual_cols = set(df.columns)
missing = expected_cols - actual_cols
extra = actual_cols - expected_cols
assert len(missing) == 0, f'Missing columns: {missing}'
if extra:
    print(f'Extra columns (will be ignored): {extra}')

# Verify target
assert set(df[TARGET_COL].unique()) == {0, 1}, f'Unexpected target values: {df[TARGET_COL].unique()}'

# Verify no NaN in features or target
assert df[FEATURE_COLS + [TARGET_COL]].isnull().sum().sum() == 0, 'Found NaN in features or target!'

# Verify no duplicate prediction units
dup_count = df.duplicated(subset=['session_id', 'visitorid', 'itemid', 'prediction_timestamp']).sum()
assert dup_count == 0, f'Found {dup_count} duplicate prediction units!'

# Check for leakage: converted column should be perfectly anti-correlated with label
if 'converted' in df.columns:
    leakage_check = (df['converted'].astype(int) + df['label']).unique()
    assert set(leakage_check) == {1}, 'WARNING: converted column is not perfectly anti-correlated with label!'
    print('Leakage check passed: converted column is perfectly anti-correlated with label.')

# Label distribution
label_dist = df[TARGET_COL].value_counts()
label_pct = df[TARGET_COL].value_counts(normalize=True) * 100
print(f'\nLabel Distribution:')
print(f'  Abandoned (1): {label_dist[1]:,} ({label_pct[1]:.1f}%)')
print(f'  Converted (0): {label_dist[0]:,} ({label_pct[0]:.1f}%)')
print(f'  Imbalance ratio: {label_dist[1]/label_dist[0]:.2f}:1')

# Timestamp range
ts_min = pd.to_datetime(df[TIMESTAMP_COL].min(), unit='ms')
ts_max = pd.to_datetime(df[TIMESTAMP_COL].max(), unit='ms')
print(f'\nTemporal range: {ts_min} to {ts_max}')
print(f'\nValidation PASSED.')
```

## 3. Chronological Train / Validation / Test Split

```python
# Sort by prediction_timestamp for chronological split
df = df.sort_values(by=TIMESTAMP_COL).reset_index(drop=True)

# Split: 60% train, 20% validation, 20% test (chronological)
n = len(df)
train_end = int(n * 0.60)
val_end = int(n * 0.80)

df_train = df.iloc[:train_end].copy()
df_val = df.iloc[train_end:val_end].copy()
df_test = df.iloc[val_end:].copy()

# Verify no temporal leakage
assert df_train[TIMESTAMP_COL].max() <= df_val[TIMESTAMP_COL].min(), 'Temporal leakage: train/val overlap!'
assert df_val[TIMESTAMP_COL].max() <= df_test[TIMESTAMP_COL].min(), 'Temporal leakage: val/test overlap!'

# Check for visitor leakage (informational, not blocking)
train_visitors = set(df_train['visitorid'])
val_visitors = set(df_val['visitorid'])
test_visitors = set(df_test['visitorid'])
val_leak = val_visitors & train_visitors
test_leak = test_visitors & (train_visitors | val_visitors)
print(f'Visitor overlap (train-val): {len(val_leak)} visitors ({len(val_leak)/len(val_visitors)*100:.1f}% of val)')
print(f'Visitor overlap (train+val-test): {len(test_leak)} visitors ({len(test_leak)/len(test_visitors)*100:.1f}% of test)')
print('Note: Some visitor overlap is expected in chronological splits. The model does not use visitorid as a feature.')

# Prepare X, y
X_train = df_train[FEATURE_COLS]
y_train = df_train[TARGET_COL]
X_val = df_val[FEATURE_COLS]
y_val = df_val[TARGET_COL]
X_test = df_test[FEATURE_COLS]
y_test = df_test[TARGET_COL]

print(f'\nSplit sizes:')
print(f'  Train: {len(X_train):,} ({len(X_train)/n*100:.1f}%) | label=1: {y_train.sum()} ({y_train.mean()*100:.1f}%)')
print(f'  Val:   {len(X_val):,} ({len(X_val)/n*100:.1f}%) | label=1: {y_val.sum()} ({y_val.mean()*100:.1f}%)')
print(f'  Test:  {len(X_test):,} ({len(X_test)/n*100:.1f}%) | label=1: {y_test.sum()} ({y_test.mean()*100:.1f}%)')

# Temporal boundaries
for name, subset in [('Train', df_train), ('Val', df_val), ('Test', df_test)]:
    t0 = pd.to_datetime(subset[TIMESTAMP_COL].min(), unit='ms')
    t1 = pd.to_datetime(subset[TIMESTAMP_COL].max(), unit='ms')
    print(f'  {name} period: {t0} to {t1}')
```

## 4. Feature Scaling

```python
# Fit scaler on training data ONLY
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

print('Scaler fitted on training data.')
print(f'Feature means (train): {np.round(scaler.mean_, 2)}')
print(f'Feature stds (train): {np.round(scaler.scale_, 2)}')
```

## 5. Train & Compare Models

```python
# Define models
models = {
    'Logistic Regression': LogisticRegression(
        class_weight='balanced', max_iter=1000, random_state=42, solver='lbfgs'
    ),
    'Random Forest': RandomForestClassifier(
        n_estimators=200, class_weight='balanced', random_state=42, n_jobs=-1,
        max_depth=15, min_samples_leaf=5
    ),
    'XGBoost': XGBClassifier(
        n_estimators=200, max_depth=6, learning_rate=0.1,
        scale_pos_weight=y_train.value_counts()[0] / y_train.value_counts()[1],
        random_state=42, eval_metric='logloss', use_label_encoder=False, n_jobs=-1
    )
}

# Train all models and collect validation results
results = {}
trained_models = {}

for name, model in models.items():
    print(f'\n=== Training {name} ===')
    model.fit(X_train_scaled, y_train)
    trained_models[name] = model

    # Predict on validation set
    y_val_prob = model.predict_proba(X_val_scaled)[:, 1]
    y_val_pred = (y_val_prob >= 0.5).astype(int)

    # Metrics
    roc = roc_auc_score(y_val, y_val_prob)
    pr_auc_val = average_precision_score(y_val, y_val_prob)
    f1 = f1_score(y_val, y_val_pred)
    prec = precision_score(y_val, y_val_pred)
    rec = recall_score(y_val, y_val_pred)

    results[name] = {
        'roc_auc': roc, 'pr_auc': pr_auc_val, 'f1': f1,
        'precision': prec, 'recall': rec,
        'y_val_prob': y_val_prob, 'y_val_pred': y_val_pred
    }

    print(f'  ROC-AUC:   {roc:.4f}')
    print(f'  PR-AUC:    {pr_auc_val:.4f}')
    print(f'  F1:        {f1:.4f}')
    print(f'  Precision: {prec:.4f}')
    print(f'  Recall:    {rec:.4f}')
    print(classification_report(y_val, y_val_pred, target_names=['Converted', 'Abandoned']))

# Comparison table
comparison_df = pd.DataFrame({
    name: {k: v for k, v in r.items() if k not in ('y_val_prob', 'y_val_pred')}
    for name, r in results.items()
}).T.round(4)
print('\n=== Model Comparison (Validation Set) ===')
print(comparison_df.to_string())
```

## 6. Visualization — Model Comparison

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

colors = {'Logistic Regression': '#1f77b4', 'Random Forest': '#ff7f0e', 'XGBoost': '#2ca02c'}

# 1. ROC Curves
for name, r in results.items():
    fpr, tpr, _ = roc_curve(y_val, r['y_val_prob'])
    axes[0].plot(fpr, tpr, label=f"{name} (AUC={r['roc_auc']:.3f})", color=colors[name])
axes[0].plot([0,1],[0,1], 'k--', alpha=0.3)
axes[0].set_xlabel('False Positive Rate')
axes[0].set_ylabel('True Positive Rate')
axes[0].set_title('ROC Curves (Validation)')
axes[0].legend(loc='lower right')

# 2. PR Curves
for name, r in results.items():
    prec_arr, rec_arr, _ = precision_recall_curve(y_val, r['y_val_prob'])
    axes[1].plot(rec_arr, prec_arr, label=f"{name} (AP={r['pr_auc']:.3f})", color=colors[name])
baseline_rate = y_val.mean()
axes[1].axhline(y=baseline_rate, color='k', linestyle='--', alpha=0.3, label=f'Baseline ({baseline_rate:.2f})')
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precision')
axes[1].set_title('Precision-Recall Curves (Validation)')
axes[1].legend(loc='lower left')

# 3. Metric comparison bar chart
metrics_to_plot = ['roc_auc', 'pr_auc', 'f1', 'precision', 'recall']
x = np.arange(len(metrics_to_plot))
width = 0.25
for i, (name, r) in enumerate(results.items()):
    vals = [r[m] for m in metrics_to_plot]
    axes[2].bar(x + i*width, vals, width, label=name, color=colors[name])
axes[2].set_xticks(x + width)
axes[2].set_xticklabels(['ROC-AUC', 'PR-AUC', 'F1', 'Precision', 'Recall'])
axes[2].set_ylim(0, 1)
axes[2].set_title('Model Comparison (Validation)')
axes[2].legend()

plt.tight_layout()
plt.savefig(f'{ARTIFACTS_DIR}/model_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 7. Select Best Model & Optimize Threshold

```python
# Select best model based on PR-AUC (more relevant for imbalanced data)
best_name = max(results, key=lambda k: results[k]['pr_auc'])
best_model = trained_models[best_name]
best_result = results[best_name]
print(f'Best model by PR-AUC: {best_name} (PR-AUC={best_result["pr_auc"]:.4f})')

# Optimize threshold on VALIDATION set using F1
y_val_prob = best_result['y_val_prob']
thresholds = np.arange(0.1, 0.9, 0.01)
f1_scores = []
for t in thresholds:
    y_pred_t = (y_val_prob >= t).astype(int)
    f1_scores.append(f1_score(y_val, y_pred_t))

best_threshold = thresholds[np.argmax(f1_scores)]
best_f1 = max(f1_scores)

print(f'\nOptimal threshold (max F1 on validation): {best_threshold:.2f}')
print(f'F1 at optimal threshold: {best_f1:.4f}')

# Plot threshold vs F1
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(thresholds, f1_scores, 'b-')
ax.axvline(x=best_threshold, color='r', linestyle='--', label=f'Best threshold={best_threshold:.2f}')
ax.set_xlabel('Classification Threshold')
ax.set_ylabel('F1 Score')
ax.set_title(f'Threshold Optimization ({best_name}, Validation Set)')
ax.legend()
plt.tight_layout()
plt.savefig(f'{ARTIFACTS_DIR}/threshold_optimization.png', dpi=150, bbox_inches='tight')
plt.show()

# Confusion matrix at optimal threshold on validation
y_val_pred_opt = (y_val_prob >= best_threshold).astype(int)
cm_val = confusion_matrix(y_val, y_val_pred_opt)
fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', ax=ax,
            xticklabels=['Converted', 'Abandoned'], yticklabels=['Converted', 'Abandoned'])
ax.set_xlabel('Predicted')
ax.set_ylabel('Actual')
ax.set_title(f'Confusion Matrix — {best_name} (Validation, threshold={best_threshold:.2f})')
plt.tight_layout()
plt.savefig(f'{ARTIFACTS_DIR}/confusion_matrix_val.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 8. Final Evaluation on Test Set

```python
# Evaluate the selected model on the HELD-OUT test set
# This is done ONCE, after all model selection and threshold tuning
y_test_prob = best_model.predict_proba(X_test_scaled)[:, 1]
y_test_pred = (y_test_prob >= best_threshold).astype(int)

test_roc_auc = roc_auc_score(y_test, y_test_prob)
test_pr_auc = average_precision_score(y_test, y_test_prob)
test_f1 = f1_score(y_test, y_test_pred)
test_precision = precision_score(y_test, y_test_pred)
test_recall = recall_score(y_test, y_test_pred)

print(f'=== FINAL TEST SET EVALUATION ===')
print(f'Model: {best_name}')
print(f'Threshold: {best_threshold:.2f}')
print(f'ROC-AUC:   {test_roc_auc:.4f}')
print(f'PR-AUC:    {test_pr_auc:.4f}')
print(f'F1:        {test_f1:.4f}')
print(f'Precision: {test_precision:.4f}')
print(f'Recall:    {test_recall:.4f}')
print()
print(classification_report(y_test, y_test_pred, target_names=['Converted', 'Abandoned']))

# Test confusion matrix
cm_test = confusion_matrix(y_test, y_test_pred)
fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Oranges', ax=ax,
            xticklabels=['Converted', 'Abandoned'], yticklabels=['Converted', 'Abandoned'])
ax.set_xlabel('Predicted')
ax.set_ylabel('Actual')
ax.set_title(f'Confusion Matrix — {best_name} (TEST, threshold={best_threshold:.2f})')
plt.tight_layout()
plt.savefig(f'{ARTIFACTS_DIR}/confusion_matrix_test.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 9. SHAP Explainability

```python
# SHAP explanations for the best model
print(f'Computing SHAP values for {best_name}...')

# Use a background sample for efficiency
background = X_train_scaled[:200]

if best_name == 'XGBoost':
    explainer = shap.TreeExplainer(best_model)
    shap_values = explainer.shap_values(X_val_scaled)
elif best_name == 'Random Forest':
    explainer = shap.TreeExplainer(best_model)
    shap_values = explainer.shap_values(X_val_scaled)
    # For binary classification RF, shap_values is a list [class0, class1]
    if isinstance(shap_values, list):
        shap_values = shap_values[1]  # Use class 1 (Abandoned)
else:
    explainer = shap.LinearExplainer(best_model, background)
    shap_values = explainer.shap_values(X_val_scaled)

print('SHAP values computed.')

# Summary plot
plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X_val_scaled, feature_names=FEATURE_COLS, show=False)
plt.title(f'SHAP Feature Importance — {best_name}')
plt.tight_layout()
plt.savefig(f'{ARTIFACTS_DIR}/shap_summary.png', dpi=150, bbox_inches='tight')
plt.show()

# Bar plot (global importance)
plt.figure(figsize=(10, 5))
shap.summary_plot(shap_values, X_val_scaled, feature_names=FEATURE_COLS, plot_type='bar', show=False)
plt.title(f'SHAP Global Feature Importance — {best_name}')
plt.tight_layout()
plt.savefig(f'{ARTIFACTS_DIR}/shap_importance_bar.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Feature importance from SHAP (mean absolute SHAP value)
shap_importance = pd.DataFrame({
    'feature': FEATURE_COLS,
    'mean_abs_shap': np.abs(shap_values).mean(axis=0)
}).sort_values('mean_abs_shap', ascending=False)

print('\n=== Feature Importance (Mean |SHAP|) ===')
for _, row in shap_importance.iterrows():
    print(f"  {row['feature']:<28} {row['mean_abs_shap']:.4f}")

# Individual prediction explanation (first 3 in validation)
print('\n=== Sample Individual Explanations ===')
for i in range(min(3, len(X_val_scaled))):
    print(f'\nSample {i+1} (actual={y_val.iloc[i]}, predicted_prob={y_val_prob[i]:.3f}):')
    sv = shap_values[i]
    feat_contribs = sorted(zip(FEATURE_COLS, sv), key=lambda x: abs(x[1]), reverse=True)
    for feat, val in feat_contribs[:5]:
        direction = 'toward Abandoned' if val > 0 else 'toward Converted'
        print(f'  {feat:<28} SHAP={val:+.4f} ({direction})')
```

## 10. Export Deployment Artifacts

```python
import sklearn
import xgboost

# 1. Save the trained model
model_path = f'{ARTIFACTS_DIR}/model.joblib'
joblib.dump(best_model, model_path)
print(f'Model saved: {model_path}')

# 2. Save the scaler
scaler_path = f'{ARTIFACTS_DIR}/scaler.joblib'
joblib.dump(scaler, scaler_path)
print(f'Scaler saved: {scaler_path}')

# 3. Save feature importance from SHAP
feature_importance = shap_importance.to_dict(orient='records')

# 4. Save model metadata / manifest
manifest = {
    'model_version': '1.0.0',
    'model_type': best_name,
    'training_date': datetime.utcnow().isoformat() + 'Z',
    'dataset': 'RetailRocket',
    'total_samples': len(df),
    'split': {
        'method': 'chronological',
        'train_size': len(X_train),
        'val_size': len(X_val),
        'test_size': len(X_test),
        'train_period': f"{pd.to_datetime(df_train[TIMESTAMP_COL].min(), unit='ms')} to {pd.to_datetime(df_train[TIMESTAMP_COL].max(), unit='ms')}",
        'val_period': f"{pd.to_datetime(df_val[TIMESTAMP_COL].min(), unit='ms')} to {pd.to_datetime(df_val[TIMESTAMP_COL].max(), unit='ms')}",
        'test_period': f"{pd.to_datetime(df_test[TIMESTAMP_COL].min(), unit='ms')} to {pd.to_datetime(df_test[TIMESTAMP_COL].max(), unit='ms')}"
    },
    'class_distribution': {
        'Abandoned': int(label_dist[1]),
        'Converted': int(label_dist[0])
    },
    'decision_threshold': float(best_threshold),
    'feature_names': FEATURE_COLS,
    'feature_count': len(FEATURE_COLS),
    'target_column': TARGET_COL,
    'target_mapping': {'0': 'Converted', '1': 'Abandoned'},
    'validation_metrics': {
        'roc_auc': round(float(best_result['roc_auc']), 4),
        'pr_auc': round(float(best_result['pr_auc']), 4),
        'f1': round(float(best_result['f1']), 4),
        'precision': round(float(best_result['precision']), 4),
        'recall': round(float(best_result['recall']), 4)
    },
    'test_metrics': {
        'roc_auc': round(float(test_roc_auc), 4),
        'pr_auc': round(float(test_pr_auc), 4),
        'f1': round(float(test_f1), 4),
        'precision': round(float(test_precision), 4),
        'recall': round(float(test_recall), 4)
    },
    'model_comparison': {
        name: {k: round(float(v), 4) for k, v in r.items() if k not in ('y_val_prob', 'y_val_pred')}
        for name, r in results.items()
    },
    'feature_importance': feature_importance,
    'confusion_matrix_test': cm_test.tolist(),
    'confusion_matrix_val': cm_val.tolist(),
    'package_versions': {
        'python': str(pd.io.common.sys.version.split()[0]),
        'scikit-learn': sklearn.__version__,
        'xgboost': xgboost.__version__,
        'pandas': pd.__version__,
        'numpy': np.__version__,
        'shap': shap.__version__
    }
}

manifest_path = f'{ARTIFACTS_DIR}/model_metadata.json'
with open(manifest_path, 'w') as f:
    json.dump(manifest, f, indent=2)
print(f'Manifest saved: {manifest_path}')

# 5. List all artifacts
print(f'\n=== Artifacts in {ARTIFACTS_DIR} ===')
for f_name in sorted(os.listdir(ARTIFACTS_DIR)):
    fpath = os.path.join(ARTIFACTS_DIR, f_name)
    size_kb = os.path.getsize(fpath) / 1024
    print(f'  {f_name:<35} {size_kb:.1f} KB')

print(f'\n=== TRAINING COMPLETE ===')
print(f'Selected model: {best_name}')
print(f'Decision threshold: {best_threshold:.2f}')
print(f'Test PR-AUC: {test_pr_auc:.4f}')
print(f'\nDownload the entire artifacts/ folder and place it in your project\'s artifacts/ directory.')
```
